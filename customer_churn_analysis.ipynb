# Customer Churn Analysis
EDA portfolio project using Python and pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/customer_churn.csv')
df['churn_flag'] = (df['churn']=='Yes').astype(int)
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
overall = df['churn_flag'].mean()
print(f'Overall churn rate: {overall:.1%}')
df.groupby('contract')['churn_flag'].mean().sort_values(ascending=False)
```

```python
rate=(df.groupby('contract')['churn_flag'].mean()*100).sort_values(ascending=False)
plt.figure(figsize=(8,5))
rate.plot(kind='bar')
plt.title('Churn Rate by Contract Type')
plt.ylabel('Churn rate (%)')
plt.tight_layout();
```
